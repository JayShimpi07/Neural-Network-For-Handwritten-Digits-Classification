```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data() # Loading the MNIST dataset
```

```python
len(X_train) # Checking the length of training data
```

```python
len(X_test) # Checking the length of testing data
```

```python
X_train[0].shape # Checking the shape of the first image in training data
```

```python
X_train[0] # Displaying the pixel values of the first image in training data
```

```python
plt.matshow(X_train[2]) # Visualizing the first image in training data
```

```python
y_train[2] # Displaying the label of the first image in training data
```

```python
y_train[:5] # Displaying the first 5 labels in training data
```

```python
X_train.shape # Checking the shape of training data
```

```python
X_train = X_train / 255 # Normalizing the training data
X_test = X_test / 255 # Normalizing the testing data
```

```python
X_train[0] # Displaying the pixel values of the first image in training data
```

```python
X_test[0] # Displaying the pixel values of the first image in testing data
```

```python
X_train_flattened = X_train.reshape(len(X_train),28*28) # Flattening the training images
X_train_flattened.shape
```

```python
X_test_flattened = X_test.reshape(len(X_test),28*28) # Flattening the testing images
X_test_flattened.shape
```

```python
X_train_flattened[0] # Displaying the first flattened training image
```

```python
model = keras.Sequential([
    keras.layers.Dense(10, input_shape=(784,),activation='sigmoid') # Single layer neural network with 10 neurons
])
model.compile(
    optimizer='adam', # Using Adam optimizer
    loss='sparse_categorical_crossentropy', # Using sparse categorical crossentropy loss
    metrics=['accuracy'] # Tracking accuracy metric
)
model.fit(X_train_flattened,y_train,epochs=5) # Training the model for 5 epochs
```

```python
model.evaluate(X_test_flattened,y_test) # Evaluating the model on testing data
```

```python
plt.matshow(X_test[0]) # Visualizing the first image in testing data
```

```python
y_predicted = model.predict(X_test_flattened) # Predicting the labels for testing data
y_predicted[0] # Displaying the predicted probabilities for the first test image
```

```python
np.argmax(y_predicted[0]) # Getting the predicted label for the first test image
```

```python
y_test[:5]
```

```python
y_predicted_labels = [np.argmax(i) for i in y_predicted] # Converting predicted probabilities to labels
y_predicted_labels[:5] # Displaying the first 5 predicted labels
```

```python
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels) # Creating the confusion matrix
cm
```

```python
import seaborn as sn
plt.figure(figsize = (10,7)) # Setting the figure size for better visibility
sn.heatmap(cm,annot=True,fmt='d') # Plotting the confusion matrix using seaborn
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
# Adding hidden layers to improve model performance
model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,),activation='relu'), # First hidden layer with 100 neurons
    keras.layers.Dense(10, activation='sigmoid') # Output layer with 10 neurons
])
model.compile(
    optimizer='adam', # Using Adam optimizer
    loss='sparse_categorical_crossentropy', # Using sparse categorical crossentropy loss
    metrics=['accuracy'] # Tracking accuracy metric
)
model.fit(X_train_flattened,y_train,epochs=5) # Training the model for 5 epochs
```

```python
model.evaluate(X_test_flattened,y_test) # Evaluating the model on testing data
```

```python
y_predicted = model.predict(X_test_flattened) # Predicting the labels for testing data
y_predicted_labels = [np.argmax(i) for i in y_predicted] # Converting predicted probabilities to labels
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels) # Creating the confusion matrix
plt.figure(figsize = (10,7)) # Setting the figure size for better visibility
sn.heatmap(cm,annot=True,fmt='d') # Plotting the confusion matrix
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
# Using Flatten layer instead of manual flattening
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)), # Flatten layer to convert 2D images to 1D
    keras.layers.Dense(100, activation='relu'), # First hidden layer with 100 neurons
    keras.layers.Dense(10, activation='sigmoid') # Output layer with 10 neurons
])
model.compile(
    optimizer='adam', # Using Adam optimizer
    loss='sparse_categorical_crossentropy', # Using sparse categorical crossentropy loss
    metrics=['accuracy'] # Tracking accuracy metric
)
model.fit(X_train,y_train,epochs=5) # Training the model for 5 epochs
```
